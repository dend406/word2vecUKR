```python
#pip install gensim
from gensim.models import KeyedVectors
```

```python
#Завантаження моделі 1-3 хв.
model = KeyedVectors.load_word2vec_format('fiction.lowercased.tokenized.word2vec.300d', binary=False)
```

```python
# Україна + Париж - Франція = ?
model.most_similar(positive=['україна', 'париж'], negative=['франція'], topn=1)
```

```python
model.most_similar(positive=['театр', 'екран'], negative=['споруда'], topn=1)
```

```python
model.most_similar(positive=['споруда'], negative=[], topn=10)
```

```python
model.most_similar(positive=['дівчинка', 'чоловік'], negative=['жінка'], topn=3)
```

```python
model.most_similar(positive=["іспит", "жах", "потуги", "надія"], negative=[], topn=5)
```

```python
class Test:
    def method_one(self):
        print("Called method_one")
    @staticmethod
    def method_two():
        print("Called method_two")

class T2(Test):
    @staticmethod
    def method_two():
        print("T2")

a_test = Test()
a_test.method_one()  # -> Called method_one
a_test.method_two()  # -> Called method_two

b_test = T2()
b_test.method_two()  # -> T2
print("-------------")
Test.method_two()  # -> Called method_two
T2.method_two()  # -> T2
```
